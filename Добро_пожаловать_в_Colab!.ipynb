```
import pandas as pd

# Ссылка на датасет Титаник (891 строка, 12 колонок)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Размер таблицы: (строки, колонки)
print(df.shape)
print("Титаник загружен")
```

```
# Переменные о датасете Титаник
dataset_name = "Титаник"
passenger_count = 891
average_fare = 71.28

# f-строка подставляет значения переменных в текст
print(f"Датасет: {dataset_name}, пассажиров: {passenger_count}, средний билет: {average_fare}")

# Проверем типы: целое число и дробное число
print(type(passenger_count))
print(type(average_fare))
```

```
# Цены билетов первых пасажиров (в фунтах)
fares = [7.25, 71.28, 8.05, 21.07, 13.0]

# Цикл проходит по каждому элементу списка
for price in fares:
  print(f"Билет за {price} фунтов")
```

```
# Один пассажир как словарь: ключ -> пассажир
passenger = {
    "Name": "Braund, Mr. Owen Harris",
    "Age": "22",
    "Sex": "male",
    "Pclass":  3,
    "Survived":  0,
    "Fare": 7.25
}

# Доступ к значению по ключу
print("Имя: ", passenger["Name"])
print("Возраст:", passenger["Age"])
print("Цена билета:", passenger["Fare"])
```
