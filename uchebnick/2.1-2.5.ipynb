```python
def f(x):
    return x * np.sin(x) - 1

a = 1
b = 2
eps = 0.0001

while abs(b - a) > eps:
    c = (a + b) / 2

    if f(a) * f(c) < 0:
        b = c
    else:
        a = c

x = (a + b) / 2

print("Корень:", round(x, 4))
print("Проверка:", round(f(x), 4))
```

```python
#упр 2
x = 1.2
eps = 0.00001

while True:
    x1 = 1 / np.sin(x)

    if abs(x1 - x) < eps:
        break

    x = x1

print("Корень:", round(x1, 5))
```

```python
#упр 2
def f(x):
    return x**3 + 3*x**2 - 24*x + 1

def df(x):
    return 3*x**2 + 6*x - 24

a = 0
b = 1
eps = 0.000001

while abs(b - a) > eps:
    a = a - f(a) / df(a)                 # касательная
    b = b - f(b) * (b - a) / (f(b)-f(a)) # хорда

x = (a + b) / 2

print("Корень:", round(x, 6))
print("Проверка:", round(f(x), 6))
```
